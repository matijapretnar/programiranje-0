## Zanka `for`

````python
def fakulteta(n):
    """Vrne fakulteto naravnega števila n."""
    produkt = 1
    i = 1
    while i <= n:
        produkt *= i
    return produkt
````

````python
def riemannova_vsota(f, a, b, n):
    """Vrne levo Riemannovo vsoto funkcije f na n enakih delih intervala [a, b]."""
    h = (b - a) / n
    vsota = 0
    for i in range(n):
        vsota += f(a + i * h) * h
    return vsota
````

````python
import math
````

````python
riemannova_vsota(math.sqrt, 0, 1, 100000000)
````

````python
riemannova_vsota(math.sqrt, 0, 1, 1000)
````

````python
def oceni_pi(n):
    v_krogu = 0
    for i in range(1, n + 1):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x ** 2 + y ** 2 <= 1:
            v_krogu += 1
        if i % 100000 == 0:
            print(4 * v_krogu / i)
    return 4 * v_krogu / n
````

````python
oceni_pi(10000000)
````

````python
def integral_monte_carlo(f, a, b, m, n):
    """Oceni integral funkcije f od a do b, kjer je 0 <= f(x) <= m,
    z n naključnimi točkami."""
    pod_grafom = 0
    for i in range(n):
        x = random.uniform(a, b)
        y = random.uniform(0, m)
        if y <= f(x):
            pod_grafom += 1
    return (b - a) * m * pod_grafom / n
````

````python
integral_monte_carlo(math.sqrt, 0, 1, 1, 1000)
````

````python
def oceni_volumen_krogle(n):
    """Oceni prostornino enotske krogle v petih razsežnostih z n naključnimi točkami."""
    v_krogli = 0
    for i in range(n):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        z = random.uniform(-1, 1)
        u = random.uniform(-1, 1)
        v = random.uniform(-1, 1)
        if x ** 2 + y ** 2 + z ** 2 + u ** 2 + v ** 2 <= 1:
            v_krogli += 1
    return 32 * v_krogli / n
````

````python
oceni_volumen_krogle(100000)
````

````python
def oceni_volumen_krogle_na_mrezi(n):
    """Oceni prostornino enotske krogle v petih razsežnostih z n ** 5 točkami na mreži."""
    h = 2 / n
    v_krogli = 0
    for i in range(n):
        for j in range(n):
            for k in range(n):
                for l in range(n):
                    for m in range(n):
                        x = -1 + (i + 0.5) * h
                        y = -1 + (j + 0.5) * h
                        z = -1 + (k + 0.5) * h
                        u = -1 + (l + 0.5) * h
                        v = -1 + (m + 0.5) * h
                        if x ** 2 + y ** 2 + z ** 2 + u ** 2 + v ** 2 <= 1:
                            v_krogli += 1
    return 32 * v_krogli / n ** 5
````

## Adaptivna integracija

````python
def trapez(f, a, b):
    return (b - a) * (f(a) + f(b)) / 2
````

````python
def integral(f, a, b, eps):
    """Vrne približek za integral funkcije f od a do b, na eps natančno."""
    c = (a + b) / 2
    celota = trapez(f, a, b)
    polovici = trapez(f, a, c) + trapez(f, c, b)
    if abs(celota - polovici) < eps:
        return polovici
    else:
        return integral(f, a, c, eps / 2) + integral(f, c, b, eps / 2)
````

````python
integral(math.sqrt, 0, 1, 1e-6)
````

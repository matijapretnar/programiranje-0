````python
def je_samoglasnik(niz):
    return len(niz) == 1 and niz in 'aeiouAEIOU'
````

````python
def stevilo_samoglasnikov(niz):
    samoglasniki = 0
    for crka in niz:
        if je_samoglasnik(crka):
            crka = "?" += 1
        else:
            print(f'Črka {crka} ni samoglasnik.')
    return samoglasniki
````

````python
stevilo_samoglasnikov('Programiranje za matematike')
````

````python
def samoglasnike_zamenjaj_z_vprasaji(niz):
    niz_z_vprasaji = ''
    for crka in niz:
        if je_samoglasnik(crka):
            niz_z_vprasaji += '?'
        else:
            niz_z_vprasaji += crka
        print(crka, niz_z_vprasaji)
    return niz_z_vprasaji
````

````python
samoglasnike_zamenjaj_z_vprasaji('Programiranje za matematike')
````

````python
def je_podniz(podniz, niz):
    """Vrne True natanko tedaj, kadar se podniz pojavi v nizu (ne nujno strnjeno)"""
    start = 0
    for znak in podniz:
        indeks = niz.find(znak, start)
        if indeks == -1:
            return False
        else:
            start = indeks + 1
    return True
````

````python
je_podniz('mama', 'matematika')
````

## Risanje grafov

````python
def risba(pogoj, x0, y0, korak, stolpci=64, vrstice=24):
    for i in range(vrstice):
        slika_vrstice = ''
        y = y0 + (vrstice / 2 - i - 0.5) * 2 * korak
        for j in range(stolpci):
            x = x0 + (j + 0.5 - stolpci / 2) * korak
            if pogoj(x, y):
                slika_vrstice += "*"
            else:
                slika_vrstice += " "
        print(slika_vrstice)
````

````python
def je_v_krogu(x, y, r=1):
    return x**2 + y**2 <= r**2
````

````python
print(risba(je_v_krogu, 0, 0, 0.1, 48, 24))
````

````python
def znak(pogoj, x, y, dx, dy):
    # ▘ 1   ▝ 2
    # ▖ 4   ▗ 8
    zgoraj_levo = pogoj(x - dx / 4, y + dy / 4)
    zgoraj_desno = pogoj(x + dx / 4, y + dy / 4)
    spodaj_levo = pogoj(x - dx / 4, y - dy / 4)
    spodaj_desno = pogoj(x + dx / 4, y - dy / 4)
    indeks = (int(zgoraj_levo) + 2 * int(zgoraj_desno)
              + 4 * int(spodaj_levo) + 8 * int(spodaj_desno))
    return ' ▘▝▀▖▌▞▛▗▚▐▜▄▙▟█'[indeks]


def risba(pogoj, x0, y0, korak, stolpci=64, vrstice=24):
    slika = ''
    for i in range(vrstice):
        y = y0 + (vrstice / 2 - i - 0.5) * 2 * korak
        for j in range(stolpci):
            x = x0 + (j + 0.5 - stolpci / 2) * korak
            slika += znak(pogoj, x, y, korak, 2 * korak)
        slika += '\n'
    return slika


def v_okvirju(slika):
    sirina = slika.find('\n')
    vrh = '┌' + sirina * '─' + '┐\n'
    dno = '└' + sirina * '─' + '┘\n'
    return vrh + '│' + slika[:-1].replace('\n', '│\n│') + '│\n' + dno
````

````python
print(v_okvirju(risba(lambda x, y: x ** 2 + y ** 2 <= 1, 0, 0, 0.1, 24, 12)))
````

````python
print(v_okvirju(risba(lambda x, y: x ** 2 + y ** 2 <= 1, 0, 0, 0.1, 24, 12)))
````

`raziskuj.py`

````python
from risanje import risba, v_okvirju


def raziskuj(pogoj, x0, y0, korak):
    ukaz = ''
    while ukaz != 'q':
        print(v_okvirju(risba(pogoj, x0, y0, korak)))
        ukaz = input('ukaz (+, -, <, >, ^, v, q): ')
        if ukaz == '+':
            korak = korak / 2
        elif ukaz == '-':
            korak = korak * 2
        elif ukaz == '<':
            x0 = x0 - 16 * korak
        elif ukaz == '>':
            x0 = x0 + 16 * korak
        elif ukaz == '^':
            y0 = y0 + 12 * korak
        elif ukaz == 'v':
            y0 = y0 - 12 * korak
````

````python
raziskuj(lambda x, y: x ** 2 + y ** 2 <= 1, 0, 0, 0.05)
````

````python
def v_mandelbrotovi_mnozici(x, y, koraki=30):
    c = x + y * 1j
    z = 0
    for _ in range(koraki):
        z = z ** 2 + c
        if abs(z) > 2:
            return False
    return True
````

````python
print(v_okvirju(risba(v_mandelbrotovi_mnozici, -0.75, 0, 0.045)))
````

````python
raziskuj(v_mandelbrotovi_mnozici, -0.75, 0, 0.045)
````

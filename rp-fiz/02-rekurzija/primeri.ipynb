## Fakulteta

```python
def fakulteta(n):
    if n == 0:
        return 1
    else:
        return n * fakulteta(n - 1)
```

```python
fakulteta(500)
```

## Evklidov algoritem

```python
def gcd(m, n):
    if n == 0:
        return m
    else:
       return gcd(n, m % n)
```

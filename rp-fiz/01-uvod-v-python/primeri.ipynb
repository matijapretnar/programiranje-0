## Večji primer: tarok

`verjetnost_trule.py`

````python
import math

# vseh kart je 54, vsak igralec jih dobi 12, v talonu pa jih ostane 6.
vseh_kart = 54
v_roki = 12  # če igramo v štiri
v_truli = 3

# math.comb(n, k) izračuna število kombinacij, kjer izmed n izberemo k predmetov
vseh_kombinacij = math.comb(vseh_kart, v_roki)

# za preostalih 9 kart v roki izberemo katerekoli izmed preostalih 51 kart
vseh_preostalih = vseh_kart - v_truli
preostalih_v_roki = v_roki - v_truli
kombinacij_s_trulo = math.comb(vseh_preostalih, preostalih_v_roki)

# verjetnost trule je delež vseh kombinacij, ki vsebujejo trulo
verjetnost_trule = kombinacij_s_trulo / vseh_kombinacij
````

````python
verjetnost_trule
````

````python
import math

vk = 54; r = 12; t = 3
k = math.comb(vk, r)
pk = vk - t
pr = r - t
kt = math.comb(pk, pr)
vt = kt / k
````

````python
import math

def verjetnost_kart_pri_taroku(v_roki, stevilo_iskanih):
    # vseh kart je 54, vsak igralec jih dobi 12, v talonu pa jih ostane 6.
    vseh_kart = 54

    # math.comb(n, k) izračuna število kombinacij, kjer izmed n izberemo k predmetov
    vseh_kombinacij = math.comb(vseh_kart, v_roki)

    # za preostalih 9 kart v roki izberemo katerekoli izmed preostalih 51 kart
    vseh_preostalih = vseh_kart - stevilo_iskanih
    preostalih_v_roki = v_roki - stevilo_iskanih
    iskanih_kombinacij = math.comb(vseh_preostalih, preostalih_v_roki)

    # verjetnost trule je delež vseh kombinacij, ki vsebujejo trulo
    return iskanih_kombinacij / vseh_kombinacij
````

````python
verjetnost_kart_pri_taroku(12, 3)
````

## Gnezdenje funkcij

`tarok.py`

````python
from math import factorial

def kombinacije(n, k):
    """Vrne število kombinacij k elementov iz množice z n elementi."""
    return factorial(n) // (factorial(k) * factorial(n - k))

def verjetnost_iskanih_kart(stevilo_iskanih, v_roki):
    """Vrne verjetnost, da pri taroku v roko dobimo iskane karte."""
    vseh_kart = 54
    vseh_kombinacij = kombinacije(vseh_kart, v_roki)
    vseh_preostalih = vseh_kart - stevilo_iskanih
    preostalih_v_roki = v_roki - stevilo_iskanih
    iskanih_kombinacij = kombinacije(vseh_preostalih, preostalih_v_roki)
    return iskanih_kombinacij / vseh_kombinacij
````

````python
verjetnost_iskanih_kart(3, 12)
````

````python
verjetnost_iskanih_kart(3, 16)
````

````python
verjetnost_iskanih_kart(1, 6)
````

## Kaj nam še manjka?

````python
def fakulteta(n):
    if n == 0:
        return 1
    else:
        return n * fakulteta(n - 1)
````

````python
fakulteta(6)
````

````python
fakulteta(10)
````

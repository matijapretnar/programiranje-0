```python
def najmanjsi_element(seznam):
    """Vrne najmanjši element v danem seznamu. Če ga ni, vrne None"""
    if len(seznam) == 0:
        return
    najmanjsi_do_zdaj = seznam[0]
    for element in seznam:
        if element < najmanjsi_do_zdaj:
            najmanjsi_do_zdaj = element
    return najmanjsi_do_zdaj
```

```python
najmanjsi_element([10, 2, 4000, 300])
```

## Funkciji `enumerate` in `zip`

```python
def vrednost_polinoma(polinom, x):
    vsota = 0
    for i in range(len(polinom)):
        koef = polinom[i]
        vsota += koef * x ** i
    return vsota
```

```python
vrednost_polinoma([3, 0, 1], 1)
```

```python
vrednost_polinoma([3, 0, 1], 2)
```

```python
def vrednost_polinoma(polinom, x):
    vsota = 0
    i = 0
    for koef in polinom:
        vsota += koef * x ** i
        i += 1
    return vsota
```

```python
def napacna_vrednost_polinoma(polinom, x):
    vsota = 0
    for koef in polinom:
        i = polinom.index(koef)
        vsota += koef * x ** i
    return vsota
```

```python
vrednost_polinoma([0, 2, 0, 2], 3)
```

```python
napacna_vrednost_polinoma([0, 2, 0, 2], 3)
```

```python
for x in enumerate('abc'):
    print(x)
```

```python
for i, x in enumerate('abc'):
    print(i, x)
```

```python
def vrednost_polinoma(polinom, x):
    vsota = 0
    for i, koef in enumerate(polinom):
        vsota += koef * x ** i
    return vsota
```

```python
for x in zip('xyz', [10, 20, 30], [4, 5, 6]):
    print(x)
```

```python
for x in zip('xyz', [10, 20, 30, 40]):
    print(x)
```

```python
def skalarni_produkt(vektor1, vektor2):
    """Vrne skalarni produkt dveh vektorjev iste dimenzije."""
    assert len(vektor1) == len(vektor2)
    vsota = 0
    for x1, x2 in zip(vektor1, vektor2):
        vsota += x1 * x2
    return vsota
```

```python
skalarni_produkt([1, -2, 5], [-2, 5, 2])
```

```python
def delitelji(n):
    seznam = []
    for d in range(1, n + 1):
        if n % d == 0:
            seznam.append(d)
    return seznam
```

```python
def fibonaccijeva_stevila(n):
    fib = [0, 1]
    for _ in range(n - 2):
        fib.append(fib[-2] + fib[-1])
    return fib
```

## Izpeljani seznami

```python
[2 * n for n in range(1, 10)]
```

```python
potence = [2 ** n for n in range(10)]
```

```python
potence
```

```python
[n - 1 for n in potence]
```

```python
[int(stevka) for stevka in str(3141592)]
```

```python
[2 * n for n in range(1, 10) if n % 3 == 1]
```

```python
def delitelji(n):
    return [d for d in range(1, n + 1) if n % d == 0]
```

```python
delitelji(60)
```

```python
all([x % 2 == 0 for x in [4, 8, 15, 16]])
```

```python
[n for n in range(2, 50) if len(delitelji(n)) == 2]
```

```python
[n for n in range(1, 10000) if sum(delitelji(n)) == 2 * n]
```
